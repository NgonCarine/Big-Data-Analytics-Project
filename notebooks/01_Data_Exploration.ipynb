# 3. KPI-Berechnung

```python
# Gesamtumsatz (Total Sales) : Der Gesamtumsatz gibt den gesamten Verkaufswert aller Bestellungen im Datensatz an. 
# Er dient als zentrale Kennzahl zur Bewertung der Unternehmensleistung

total_sales = df["Sales"].sum()

print(f"Gesamtumsatz: {total_sales:.2f} $")
```

```python
# Gesamtgewinn (Total Profit) : Der Gesamtgewinn zeigt, wie profitabel das Unternehmen insgesamt arbeitet. 
# Positive Werte bedeuten Gewinn, negative Werte würden auf Verluste hinweisen.

total_profit = df["Profit"].sum()

print(f"Gesamtgewinn: {total_profit:.2f} $")
```

```python
# Verkaufte Produkte : Die verkaufte Stückzahl zeigt das gesamte Verkaufsvolumen des Unternehmens und ergänzt die Umsatzanalyse.

total_quantity = df["Quantity"].sum()

print(f"Verkaufte Produkte: {total_quantity}")
```

```python
# Gewinnmarge : Die Gewinnmarge beschreibt den prozentualen Anteil des Gewinns am Gesamtumsatz. 
# Sie ist eine wichtige Kennzahl zur Bewertung der wirtschaftlichen Effizienz.

profit_margin = (df["Profit"].sum() / df["Sales"].sum()) * 100

print(f"Gewinnmarge: {profit_margin:.2f} %")
```

# 4. Aggregationslogik

## Umsatz und Gewinn nach Kategorie

```python
# Die Verkaufsdaten wurden nach Produktkategorie aggregiert. 
# Dadurch werden Umsatz, Gewinn und Verkaufsmenge pro Kategorie zusammengefasst. 
# Diese Aggregation bildet die Grundlage zur Beantwortung der ersten Business-Frage und dient gleichzeitig als Basis für das Tableau-Dashboard.

category_summary = (
    df.groupby("Category")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Quantity=("Quantity", "sum")
      )
      .round(2)
)

category_summary
```
